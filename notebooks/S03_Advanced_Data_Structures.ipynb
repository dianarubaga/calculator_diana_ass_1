## 1. Lists: ordered and mutable

Lists preserve order and can change. Useful operations include `append`, `extend`, `insert`, `remove`, `pop`, indexing, slicing, `len`, `sum`, `min`, `max`, `sorted` and `reversed`.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
amounts = [18.5, 42.0, 7.5]
amounts.append(12.0)
first_two = amounts[:2]
largest = max(amounts)
ascending = sorted(amounts)
print(amounts, first_two, largest, ascending)
```

## 2. Tuples: fixed records and unpacking

## 3. Sets: uniqueness and membership

Sets store unique values. Use `add`, `discard`, union `|`, intersection `&`, difference `-`, `issubset()` and membership with `in`. Sets do not guarantee display order.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
required = {"python", "sql", "git"}
team = {"python", "excel", "git", "python"}
missing = required - team
shared = required & team
all_skills = required | team
print(missing, shared, all_skills)
```

## 4. Dictionaries: labelled data

Dictionaries map unique keys to values. Use direct indexing for required keys, `get()` for optional keys, `keys()`, `values()`, `items()`, `update()`, `setdefault()` and `pop()`.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
customer = {"id":"C01","name":"Ana","segment":"premium"}
print(customer["name"])
print(customer.get("country","unknown"))
customer.update({"country":"ES"})
for key,value in customer.items():
    print(key,value)
```

## 5. Nested collections

Real data often combines collections: a list of dictionaries or a dictionary whose values are lists. Traverse one level at a time and keep every record shape consistent.


### Worked example

Predict the result first. Then run the cell and change one input.


```python
orders = [
    {"id":"O1","customer":"C1","items":[12.0,8.0]},
    {"id":"O2","customer":"C2","items":[20.0]},
]
for order in orders:
    order["total"] = sum(order["items"])
print(orders)
```

### Guided exercise 5

Build a dictionary indexed by contact ID from a list of contacts. Then find all IDs tagged as clients.


```python
contacts = [
    {"id":"C1","name":"Ana","tags":{"client","newsletter"}},
    {"id":"C2","name":"Luis","tags":{"supplier"}},
]
# TODO: by_id dictionary
by_id = {contact["id"]: contact for contact in contacts}
# TODO: client_ids list
client_ids = [contact["id"] for contact in contacts if "client" in contact["tags"]]
print(contacts)
print(by_id)
print(client_ids)
```

## Final guided challenge

Create a grade tracker: dictionary indexed by student ID, list of grades per student, a set of subjects, average calculation, and ranking that safely handles students with no grades.

Build it in small steps and keep every previous check passing.


```python
students = {
    "S1":{"name":"Ana","grades":[8,9],"subjects":{"python","data"}},
    "S2":{"name":"Luis","grades":[],"subjects":{"python"}},
}
# TODO: average(student_id) -> float or None
def average(student_id):
    student = students.get(student_id)
    if student and student["grades"]:
        return sum(student["grades"]) / len(student["grades"])
    return None
# TODO: ranked list excluding students without grades
ranked = sorted(
    [student for student in students.values() if student["grades"]],
    key=lambda student: sum(student["grades"]) / len(student["grades"]),
    reverse=True
)
# TODO: unique subjects taught
unique_subjects = {subject for student in students.values() for subject in student["subjects"]}
# TODO: checks
print(average("S1"))
print(average("S2"))
print(ranked)
print(unique_subjects)

```
